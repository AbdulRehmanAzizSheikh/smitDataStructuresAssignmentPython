## Create a dictionary that manages records of two patients.

Each patient should have:

    Patient ID

    Medicines (a nested dictionary of medicine names and their doses)

### Example:

    Patient 1: ID 501, Name Bilal, Age 45, Disease Diabetes, Doctor Dr. Ayesha
    Medicines: Metformin 500mg, Insulin 10 units
    Patient 2: ID 502, Name Hina, Age 32, Disease Flu, Doctor Dr. Usman

## Create a nested dictionary for a company that stores three employees.

Each employee should have:

    Employee ID
    Name
    Department
    Salary
    Skills (a list of skills inside the dictionary)

Employees:

    ID 201: Name Ali, Department IT, Salary 75000, Skills: Python, SQL, HTML
    ID 202: Name Fatima, Department HR, Salary 60000, Skills: Communication, Recruitment
    ID 203: Name Zain, Department Finance, Salary 80000, Skills: Excel, Accounting

### Tasks:

    Print the complete employee database.
    Print only the skills of employee 201.
    Add a new skill "JavaScript" for Ali.
    Increase Fatima’s salary to 65000.
    Add a new employee: ID 204, Name Hina, Department Marketing, Salary 55000, Skills: SEO, Content Writing.
    Remove employee 203.
    Print the final database.

```python
company = {
    201: {"Employee ID": 201, "Name": "Ali", "Department": "IT",
          "Salary": 75000, "Skills": ["Python", "SQL", "HTML"]},
    202: {"Employee ID": 202, "Name": "Fatima", "Department": "HR",
          "Salary": 60000, "Skills": ["Communication", "Recruitment"]},
    203: {"Employee ID": 203, "Name": "Zain", "Department": "Finance",
          "Salary": 80000, "Skills": ["Excel", "Accounting"]}
}

print("Complete employee database:")
print(company)
print("Skills of employee 201:", company[201]["Skills"])

company[201]["Skills"].append("JavaScript")
company[202]["Salary"] = 65000
company[204] = {"Employee ID": 204, "Name": "Hina", "Department": "Marketing",
                "Salary": 55000, "Skills": ["SEO", "Content Writing"]}
del company[203]

print("Final database:")
print(company)
```

## Common Subjects Between Two Student
Create a set :

    Ali’s subjects: Math, Science, English, Computer, Urdu

    Sara’s subjects: Science, English, Islamiat, Computer, Arts
## Find and print:

    Subjects that both students study (common subjects)
    All subjects taken by either Ali or Sara (combined unique subjects)
    Subjects that only Ali studies
    Subjects that only Sara studies

```python
ali = {"Math", "Science", "English", "Computer", "Urdu"}
sara = {"Science", "English", "Islamiat", "Computer", "Arts"}

print("Common subjects:", ali & sara)
print("All subjects:", ali | sara)
print("Only Ali:", ali - sara)
print("Only Sara:", sara - ali)
```
